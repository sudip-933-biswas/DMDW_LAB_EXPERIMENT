```python
import numpy as np
import pandas as pd
Study_hour=[1,2,3,4,5]
marks=[40,50,60,70,80]
x_mean=np.average(Study_hour)
y_mean=np.average(marks)
print(x_mean)
print(y_mean)
xi_minus_x_mean=[]
yi_minus_y_mean=[]
mul=[]
for i in range(len(Study_hour)):
    xi_minus_x_mean.append(Study_hour[i]-x_mean)
for i in range(len(marks)):
    yi_minus_y_mean.append(marks[i]-y_mean)
for i in range(len(xi_minus_x_mean)):
    mul.append(xi_minus_x_mean[i]*yi_minus_y_mean[i])
sum=np.sum(mul)
cor=sum/(len(Study_hour)-1)
print(f"covariance is:{cor}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

Study_hour = [1, 2, 3, 4, 5]
marks = [40, 50, 60, 70, 80]

x_mean = np.average(Study_hour)
y_mean = np.average(marks)

print(x_mean)
print(y_mean)

cofactor = []
x_square = []
y_square = []

for i in range(len(Study_hour)):

    x_diff = Study_hour[i] - x_mean
    y_diff = marks[i] - y_mean

    cofactor.append(x_diff * y_diff)

    x_square.append(x_diff ** 2)
    y_square.append(y_diff ** 2)

numerator = np.sum(cofactor)

denominator = np.sqrt(np.sum(x_square) * np.sum(y_square))

correlation = numerator / denominator
plt.figure(figsize=(8, 6))
print("Correlation is:", correlation)
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
