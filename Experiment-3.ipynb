```python
#Dissimilarity Matrix
import numpy as np #23/9/26
import pandas as pd
data={
    'size':['Small','Midium','Small','Large','Large'],
    'Color':['Red','Red','Blue','Green','yellow']
}
df=pd.DataFrame(data,index=['f1','f2','f3','f4','f5'])#(DataFrame)it keep data in tabular format
print(df)
n=len(df)
dissimilarity_matrix=np.zeros((n,n))
for i in range(n):
  for j in range(n):
    #count no of mismatchesnattributes
    mismatches=np.sum(df.iloc[i] != df.iloc[j])
    total_attributes=df.shape[1]
    dissimilarity_matrix[i,j]=(mismatches/total_attributes)
  dissimilarity_df=pd.DataFrame(
      dissimilarity_matrix,
      index=df.index,
      columns=df.index
  )
print("\nDissimilarity matrix:")
print(dissimilarity_df)
```

```python
import numpy as np
import pandas as pd
data={
    'size':['Small','Midium','Small','Large','Large'],
    'Color':['Red','Red','Blue','Green','yellow']
}
df=pd.DataFrame(data,index=['f1','f2','f3','f4','f5'])
print("Original DataSet")
print(df)
size_mapping={'Small':1,'Midium':2,'Large':3}
df['Size_Rank']=df['size'].map(size_mapping)
print("\nSize after Ranking:",df[['size','Size_Rank']])
min_rank=df['Size_Rank'].min()
max_rank=df['Size_Rank'].max()
df['Size_Normalized']=(df['Size_Rank']-1)/(max_rank - 1)
print("\nNormalized Size: ")
print(df[['size','Size_Normalized']])
n = len(df)

dissimilarity_matrix = np.zeros((n,n))

for i in range(n):
    for j in range(n):

        size_dissimilarity = abs(
            df.iloc[i]['Size_Normalized'] -
            df.iloc[j]['Size_Normalized']
        )

        dissimilarity_matrix[i,j] = size_dissimilarity

dissimilarity_df = pd.DataFrame(
    dissimilarity_matrix,
    index=df.index,
    columns=df.index
)

print("Dissimilarity Matrix:")
print(dissimilarity_df)
```

```python
# For Binary Attribute

import numpy as np
import pandas as pd

binary_data = pd.DataFrame({
    'Feature A': [1, 1, 0, 0],
    'Feature B': [1, 0, 1, 0],
    'Feature C': [0, 1, 1, 0]
})

print("\nBinary DataSet")
print(binary_data)

# ✅ Number of objects
n = len(binary_data)

matrix = np.zeros((n, n))

for i in range(n):
    for j in range(n):

        a = b = c = 0

        # Compare all binary attributes
        for k in range(binary_data.shape[1]):

            x = binary_data.iloc[i, k]
            y = binary_data.iloc[j, k]

            if x == 1 and y == 1:
                a += 1

            elif x == 1 and y == 0:
                b += 1

            elif x == 0 and y == 1:
                c += 1

        # Calculate denominator AFTER checking all attributes
        denominator = a + b + c

        if denominator == 0:
            matrix[i, j] = 0
        else:
            matrix[i, j] = (b + c) / denominator


print("\nBinary Attribute Dissimilarity Matrix:")

print(
    pd.DataFrame(
        matrix,
        index=['object 1', 'object 2', 'object 3', 'object 4'],
        columns=['object 1', 'object 2', 'object 3', 'object 4']
    )
)
```
