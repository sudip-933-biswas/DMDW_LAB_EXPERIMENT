```python
import pandas as pd
import numpy as np

data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [10, 20, 30, 40, 50]
}

marks = data["Marks"]

min_value = min(marks)
max_value = max(marks)

min_max = []
for x in marks:
    normalized = (x - min_value) / (max_value - min_value)
    min_max.append(normalized)

data["Min_Max_Normalized"] = min_max
data=pd.DataFrame(data)

print(data)
```

```python
import statistics
import pandas as pd
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [10, 20, 30, 40, 50]
}


marks = data["Marks"]

mean_value = sum(marks) / len(marks)

std_value = statistics.pstdev(marks)

z_score = []

for x in marks:
    normalized = (x - mean_value) / std_value
    z_score.append(normalized)

data["Z_Score_Normalized"] = z_score
data=pd.DataFrame(data)
print(data)
```
